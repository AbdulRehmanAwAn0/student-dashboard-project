# NYSE Listed Companies — Exploratory Data Analysis

Quick EDA notebook accompanying the Streamlit dashboard.

```python
import sys, os
sys.path.append(os.path.abspath('..'))

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from filters import load_data
from charts import apply_theme

apply_theme()
df = load_data('../data/nyse-Dashboard listed.csv')
df.head()
```

```python
print('Shape:', df.shape)
print('\nMissing values:')
print(df.isna().sum())
df.describe(include='all').T.head(20)
```

```python
df['Security Type'].value_counts().plot(kind='barh', figsize=(8,5), color='#2E86AB')
plt.title('Security-Type Distribution'); plt.xlabel('Count'); plt.tight_layout()
```

```python
sns.heatmap(df[['Symbol Length','Name Length','Word Count','Coupon Rate','Listing Year']].corr(),
            annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Correlation Matrix')
```
